## Dimensionality Reduction

The objective is to select the best components (attributes) for algorithm training through the analysis of correlations between variables.

### Principal Component Analysis (PCA)

Feature Selection: Selects the best attributes and uses them without transformations.

Feature Extraction: Identifies relationships among the best attributes and creates new ones.

It is an unsupervised learning algorithm.

Applies to linearly separable data.

```python
pca = PCA(n_components=3)
```

# TRAINING AND TEST SET

Parameters for train_test_split:



* arrays: Names of predictor and target attributes.
* test_size: Size in percentage of test data. Default is none.
* train_size: Size in percentage of training data. Default is none.
* random_state: Naming a random state.
* shuffle: Random data shuffling. When associated with random_state, the same shuffling occurs every time. Default is True.
* stratify: Possibility of stratified data splitting. Default is None (in this case, the proportion is maintained, i.e., if there are 30% zeros and 70% ones in the dataframe, the split into training and testing will maintain this ratio).

**Adjustment Control**

num_leaves: defines the number of leaves to be formed in a tree. There is no direct relationship between num_leaves and max_depth, and therefore, the two should not be linked to each other.

max_depth: specifies the maximum depth or level to which the tree can grow.

**Speed Control**

learning_rate: learning rate determines the impact of each tree on the final result.

max_bin: A smaller value for max_bin greatly reduces processing time by discretizing feature values into discrete bins, which is computationally cheaper.

**Precision Control**

num_leaves: a high value produces deeper trees with higher precision but leads to overfitting.

max_bin: high values have a similar effect to increasing the value of num_leaves and also slow down the training process.

## **Grid Search (Gradient Boosting Classifier)**

```python
# Creating the algorithm
gbm = GradientBoostingClassifier(random_state=0)
```

```python
# Defining the hyperparameter space
param_grid = dict(
    n_estimators=[20, 50, 100, 250],
    learning_rate=[0.05, 0.1, 0.5],
    max_depth=[1, 2, 3, 4, 5],
)
```

```python
print('Number of Hyperparameter Combinations: ',
      len(param_grid['n_estimators']) * len(param_grid['learning_rate']) * len(param_grid['max_depth']))
```

```python
# Configuring the search with Grid search
grid_search = GridSearchCV(gbm, param_grid, scoring='roc_auc', cv=4)
```

```python
# Configuring the best hyperparameters
grid_search.fit(x_train, y_train)
```

```python
# Best Hyperparameters
best_hyperparameters = grid_search.best_params_
print("Best Hyperparameters:", best_hyperparameters)
```

```python
# Data relationship for all models
result_df = pd.DataFrame(grid_search.cv_results_)
print(result_df.shape)
result_df.head(3)
```
