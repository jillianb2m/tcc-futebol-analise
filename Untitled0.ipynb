```python
import time
import threading

pontos = 0
enter_pressionado = False

def esperar_enter():
    global enter_pressionado
    input()
    enter_pressionado = True

dezena = int(input("Qual dezena você quer jogar? (ex: 20, 30, 40): "))

for numero in range(dezena, dezena + 10):

    enter_pressionado = False
    print("Pressione ENTER quando terminar de anotar!")

    thread = threading.Thread(target=esperar_enter)
    thread.start()

    inicio = time.time()

    while time.time() - inicio < 7:
        if enter_pressionado:
            pontos += 1
            print(f"\nNúmero: {numero}")
            print("✅ Ponto para você!")
            break
        time.sleep(0.1)

    if not enter_pressionado:
        print("⏰ Tempo esgotado!")

print(f"\n🏆 Pontuação final: {pontos}")
```
