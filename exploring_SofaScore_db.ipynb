```python
import pandas as pd
import glob
```

```python
arquivos = ['brasileirao_2023_completo.csv', 'brasileirao_2024_completo.csv', 'brasileirao_2025_completo.csv']
```

```python
lista_df = [pd.read_csv(f) for f in arquivos]
```

```python
df_final = pd.concat(lista_df, ignore_index=True)
df_final.head()
```

```python
df_final.describe()
```

```python
nulos_por_coluna = df_final.isnull().sum()

total_linhas = len(df_final)

resumo_nan = pd.DataFrame({
    'Total NaN': nulos_por_coluna,
    'Total Linhas': total_linhas,
    'Percentual (%)': (nulos_por_coluna / total_linhas) * 100
})

print(resumo_nan)
```
