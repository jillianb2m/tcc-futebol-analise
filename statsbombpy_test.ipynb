```python
jogo = sb.events(match_id=3773386)
jogo['type'].value_counts()
```

```python
jogador = jogo.dropna(subset=['player'])
jogador.head()
```

```python
# 3. Agrupar por Jogador e Time e calcular as métricas
resumo = jogador.groupby(['player', 'team']).agg(
    passes=('type', lambda x: (x == 'Pass').sum()),
    chutes=('type', lambda x: (x == 'Shot').sum()),
    dribles_completos=('type', lambda x: ((x == 'Dribble') & (events.loc[x.index, 'dribble_outcome'] == 'Complete')).sum()),
    desarmes=('type', lambda x: (x == 'Duel').sum()), # Aproximação de duelos defensivos
    xg_acumulado=('shot_statsbomb_xg', 'sum')
).reset_index()

# 4. Ver os top 10 jogadores com mais passes
print(resumo.sort_values(by='passes', ascending=False).head(10))
```
