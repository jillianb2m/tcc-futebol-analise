# Resumo PRÉ-PROCESSAMENTO

**alvo** = variável que se pretende atingir (tem ou não doença cardíaca) - ***Meu comentário: Resultado que será calculado pelo ML***

**previsores** = conjunto de variáveis previsoras com as variáveis categóricas transformadas em numéricas manualmente, sem escalonar. ***Meu comentário: Foram transformadas as colunas não-numéricas para numéricas (Ex.: Sex)***

**previsores_esc** = conjunto de variáveis previsoras com as variáveis categóricas transformadas em numéricas, escalonada. ***Meu comentário: Transformou todas colunas (aqui todas já estavam como numéricas) de forma que ficassem escalonadas (números próximos de zero)***

**previsores2** = conjunto de variáveis previsoras com as variáveis categóricas transformadas em numéricas pelo labelencoder. ***Meu comentário: Com base na variável previsores foi gerado automaticamente a previsores2***

**previsores3** = conjunto de variáveis previsoras transformadas pelo labelencoder e onehotencoder, sem escalonar. ***Meu comentário: Transformadas as colunas que anteriormente eram "Strings" em onehotencoder (que criou novas colunas conforme a necessidade dos valores das colunas)***

**previsores3_esc** = conjunto de variáveis previsoras transformadas pelo labelencoder e onehotencoder escalonada. ***Meu comentário: Usou a previsores3 para escalonar (ou seja, números perto de zero)***
